```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
```

```python
x=np.linspace(1,5,10)
x
y=x**5
y
```

BASIC OF MATPLOTLIB

```python
plt.plot(x,y)
plt.title('learning')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
```

```python
plt.subplot(2,2,2)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(x,y**2)
plt.subplot(2,2,1)
plt.plot(y,x)
plt.subplot(2,2,4)
plt.plot(x,y)
```

```python
plt.scatter(x,y)
```
