```python
import numpy as np

X_train = np.array([
    [0.5, 1.5],
    [1, 1],
    [1.5, 0.5],
    [3, 0.5],
    [2, 1],
    [4, 1]
])

y_train = np.array([0, 0, 0, 1, 1, 1])


def cost(x, y, w, b):
    n = x.shape[0]
    ans = 0

    for i in range(n):
        z = np.dot(x[i], w) + b
        print(z)

        sig = 1 / (1 + np.exp(-z))

        fun = -y[i] * np.log(sig) - (1 - y[i]) * np.log(1 - sig)

        ans += fun / n

    return ans

print(cost(X_train,y_train,[1,2],1))

```
