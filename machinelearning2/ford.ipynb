```python
X = df.drop(columns = ['price'])
y = df['price']
```

```python
X
```

```python
X_one_encode = pd.get_dummies(X,columns = ['model','transmission','fuelType'],drop_first = True)
```

```python
X_one_encode=X_one_encode.astype(int)
```

```python
X_one_encode
```

```python
from sklearn.preprocessing import LabelEncoder
col=['model', 'transmission', 'fuelType']
Xlabel=X
for i in col:
    le=LabelEncoder()
    Xlabel[i]=le.fit_transform(Xlabel[i].astype(str))
```

```python
Xlabel
```

```python
from sklearn.preprocessing import StandardScaler
model=StandardScaler()
columns= ['year', 'mileage', 'tax', 'mpg', 'engineSize']
X_one_encode[columns]=model.fit_transform(X_one_encode[columns])
```

```python
X_one_encode
```

```python
new_colums=['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg','engineSize']
Xlabel[new_colums]=model.fit_transform(Xlabel[new_colums])
```

```python
Xlabel
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
machine_model=LinearRegression()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_one_encode, y, test_size=0.33, random_state=42)  
```

```python
machine_model.fit(X_train,y_train)
```

```python
y_pred1=machine_model.predict(X_test)
y_pred1
```

```python
X_train, X_test, y_train, y_test = train_test_split(Xlabel, y, test_size=0.33, random_state=42) 
```

```python
machine_model.fit(X_train,y_train)
```

```python
y_pred2=machine_model.predict(X_test)
y_pred2
```

```python
r1=r2_score(y_test,y_pred1)
r1*100#  for one hot encoding
```

```python
r2=r2_score(y_test,y_pred2)
r2*100 #for label encoding
```

```python
n = X_test.shape[0]# for one hot encoding
p = X_test.shape[1]
adjusted_r1 = 1 - ((1 - r1) * (n - 1)) / (n - p - 1)
print("Adjusted R² Score:", adjusted_r1)
```

```python
n = X_test.shape[0] #for label encoding
p = X_test.shape[1]
adjusted_r2 = 1 - ((1 - r2) * (n - 1)) / (n - p - 1)
print("Adjusted R² Score:", adjusted_r2)
```
